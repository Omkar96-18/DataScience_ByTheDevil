```python
import ast
ast.literal_eval("[{'id': 16, 'name': 'Animation'}, {'id': 35, 'name': 'Comedy'}, {'id': 10751, 'name': 'Family'}]")
```

```python
def convert(obj):
    list = []
    for i in ast.literal_eval(obj):
        list.append(i['name'])

    return list
```

```python
print(convert("[{'id': 16, 'name': 'Animation'}, {'id': 35, 'name': 'Comedy'}, {'id': 10751, 'name': 'Family'}]"))
```

```python
movies['genres'] = movies['genres'].apply(convert)
```
